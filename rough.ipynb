Fwhm of a double gaussian

```python
import numpy as np
import matplotlib.pyplot as plt
from models import Models
```

```python
x = np.linspace(-10,10,300)
a1=4
a2=5
mu1=-1
mu2=1
sigma1=2
sigma2=2
m=0
c=0

G = Models.Gauss2linearbackground(x,a1,a2,mu1,mu2,sigma1,sigma2,m,c)
G1 = Models.Gauss1linearbackground(x,a1,mu1,sigma1,m,c)
G2 = Models.Gauss1linearbackground(x,a2,mu2,sigma2,m,c)

def FWHM2Gauss():
    global G
    global x
    
    max_ind = np.argmax(G)
    max_G = np.max(G)
    
    fwhm_ind_up = np.where(G>=max_G/2)[0]
    return x[fwhm_ind_up],G[fwhm_ind_up]

x_up,G_up = FWHM2Gauss()
```

```python
fig1 = plt.figure()
ax1f1 = fig1.add_subplot(1,1,1)
im1f1 = ax1f1.plot(x,G)
im2f1 = ax1f1.plot(x,G1)
im3f1 = ax1f1.plot(x,G2)

im5f1 = ax1f1.plot(x_up,G_up)



plt.show()
```
